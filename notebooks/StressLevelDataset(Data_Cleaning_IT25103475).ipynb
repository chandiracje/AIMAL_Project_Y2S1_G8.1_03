```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
null_counts = stressLvlDf.isnull().sum()

plt.figure(figsize=(10, 6))
null_counts.plot(kind='barh', color='#4C7A73')
plt.title("Missing values per Column")
plt.xlabel("Count")
plt.tight_layout()
plt.show()

print("Total missing cells:", null_counts.sum())
```

```python
#Heat Map Version
```

```python
plt.figure(figsize=(14, 6))
sns.heatmap(stressLvlDf.isnull(), cbar=False, cmap='Reds', yticklabels=False)
plt.title("Missing Value Heatmap")
plt.tight_layout()
plt.show()
```
